```python
# sql_engine: bigquery
# output_variable: df
# output_mode: table
# start _sql
_sql = """
SELECT
  COUNT(*) AS total_facts,
  COUNT(f.unique_key) AS unique_crimes,
  COUNT(d_type.SK_incident_type) AS matched_crime_types,
  COUNT(d_loc.SK_area_id) AS matched_locations,
  COUNT(d_cal.SK_police_info) AS police_info
FROM `dataengineerin-510510.DEV_CHICAGO.DM_FAIT_INCIDENT_REPORT` f
LEFT JOIN `dataengineerin-510510.DEV_CHICAGO.DM_VW_INCIDENT_TYPE` d_type
  ON f.FK_incident_type = d_type.SK_incident_type
LEFT JOIN `dataengineerin-510510.DEV_CHICAGO.DM_VW_AREA` d_loc
  ON f.FK_area_id = d_loc.SK_area_id
LEFT JOIN `dataengineerin-510510.DEV_CHICAGO.DM_VW_POLICE_INFO` d_cal
  ON f.FK_police_info = d_cal.SK_police_info;
""" # end _sql
from google.colab.sql import bigquery as _bqsqlcell
df = _bqsqlcell.run(_sql)
df
```
