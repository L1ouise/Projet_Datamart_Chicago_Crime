```python
CREATE OR REPLACE TABLE `dataengineerin-510510.DEV_CHICAGO.DM_FAIT_INCIDENT_REPORT` AS
SELECT
date AS registeredDate,
case_number,
unique_key,
arrest AS is_arrest,
domestic AS is_domestic,
year,
updated_on,
CAST(ABS(FARM_FINGERPRINT(CONCAT(IFNULL(community_area, 0), IFNULL(ward, 0), IFNULL(location, ""), IFNULL(block,"")))) AS INT64) AS FK_area_id,
CAST(ABS(FARM_FINGERPRINT(CONCAT(IFNULL(iucr, ""),IFNULL(description,"")))) AS INT64) AS FK_crime_descriprion,
CAST(ABS(FARM_FINGERPRINT(CONCAT(IFNULL(fbi_code,""), IFNULL(Primary_type,"")))) AS INT64) AS FK_incident_type,
CAST(ABS(FARM_FINGERPRINT(CONCAT(IFNULL(district, 0), IFNULL(beat, 0)))) AS INT64) AS FK_police_info
FROM `bigquery-public-data.chicago_crime.crime`
WHERE unique_key IS NOT NULL;
```
