```python
CREATE OR REPLACE TABLE dataengineerin-510510.DEV_CHICAGO.DM_VW_AREA AS
SELECT
CAST(ABS(FARM_FINGERPRINT(CONCAT(IFNULL(community_area, 0), IFNULL(ward, 0), IFNULL(location, ""), IFNULL(block,"")))) AS INT64) AS SK_area_id,
--ROW_NUMBER() OVER()  AS SK_area_id,
community_area,
ward,
block,
location,
x_coordinate,
y_coordinate
FROM bigquery-public-data.chicago_crime.crime
GROUP BY community_area, ward, block, location, x_coordinate, y_coordinate
ORDER BY community_area, ward ASC;
```

```python
CREATE OR REPLACE TABLE `dataengineerin-510510.DEV_CHICAGO.DM_VW_POLICE_INFO` AS
SELECT
CAST(ABS(FARM_FINGERPRINT(CONCAT(IFNULL(district, 0), IFNULL(beat, 0)))) AS INT64) AS SK_police_info,
--ROW_NUMBER() OVER() AS SK_police_info,
district,
beat
FROM bigquery-public-data.chicago_crime.crime
GROUP BY district, beat
ORDER BY district
ASC
```

```python
CREATE OR REPLACE TABLE `dataengineerin-510510.DEV_CHICAGO.DM_VW_INCIDENT_TYPE` AS
SELECT
CAST(ABS(FARM_FINGERPRINT(CONCAT(IFNULL(fbi_code,""), IFNULL(Primary_type,"")))) AS INT64) AS SK_incident_type,
--ROW_NUMBER() OVER() AS SK_incident_type,
fbi_code, Primary_type
FROM bigquery-public-data.chicago_crime.crime
GROUP BY Primary_type, fbi_code
ORDER BY fbi_code
ASC
```

```python
CREATE OR REPLACE TABLE `dataengineerin-510510.DEV_CHICAGO.DM_VW_CRIME_DESCRIPTION` AS
SELECT
CAST(ABS(FARM_FINGERPRINT(CONCAT(IFNULL(iucr, ""),IFNULL(description,"")))) AS INT64) AS SK_crime_descriprion,
--ROW_NUMBER() OVER() AS SK_crime_descriprion,
iucr,
description
FROM bigquery-public-data.chicago_crime.crime
GROUP BY iucr, description
ORDER BY iucr, SK_crime_descriprion, description
ASC
```
